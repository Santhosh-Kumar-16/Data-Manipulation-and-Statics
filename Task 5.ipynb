```python
# Create Profit Amount Column

# Profit Amount indicates the actual profit earned from sales.

# Profit Amount = Revenue × Profit Margin / 100
```

```python
df["Profit_Amount"] = (
    df["Revenue"] *
    df["Profit_Margin"] / 100
)

df[["Revenue",
    "Profit_Margin",
    "Profit_Amount"]].sample(5)
```

```python
# Create Stock Difference Column

# This feature shows the difference between available stock and units sold.
```

```python
df["Stock_Difference"] = (
    df["Quantity_Available"] -
    df["Units_Sold"]
)

df[["Quantity_Available",
    "Units_Sold",
    "Stock_Difference"]].sample(5)
```

```python
# # Create  Inventory Status Column

# # Categorize inventory levels based on available quantity.


# Condition:

# Low Stock → Quantity < 100
# Medium Stock → Quantity 100–200
# High Stock → Quantity > 200
```

```python
df["Inventory_Status"] = np.where(
    df["Quantity_Available"] < 100,
    "Low Stock",
    np.where(
        df["Quantity_Available"] <= 200,
        "Medium Stock",
        "High Stock"
    )
)

df[["Quantity_Available",
    "Inventory_Status"]].sample(5)
```

```python
# Create Revenue Per Unit Column

# Revenue generated for each unit sold
```

```python
df["Revenue_Per_Unit"] = (
    df["Revenue"] /
    df["Units_Sold"]
)

df[["Revenue",
    "Units_Sold",
    "Revenue_Per_Unit"]].sample(5)
```

```python
# Task 2: Transform Data

# Retail datasets often contain raw numerical and textual data that may not be easy to interpret directly.


# For example:

# Revenue values such as 8,452 or 12,340 may be difficult to analyze individually.
# Customer Satisfaction scores such as 4.5 or 3.2 may not clearly indicate customer sentiment.
# Discount percentages may need to be grouped into categories for reporting.
 

# To make data easier to understand and analyze, analysts perform data transformation.

# What is Data Transformation?

# # Data Transformation is the process of converting data from one format or structure into another format that is more useful for analysis.
```

```python
#Transform Revenue into Revenue Category

df["Revenue_Category"] = np.where(
    df["Revenue"] < 5000,
    "Low Revenue",
    np.where(
        df["Revenue"] < 10000,
        "Medium Revenue",
        "High Revenue"
    )
)

df[["Revenue", "Revenue_Category"]].head()
```

```python
#Transform Customer Satisfaction Ratings

df["Customer_Feedback"] = np.where(
    df["Customer_Satisfaction"] >= 4,
    "Satisfied",
    "Needs Improvement"
)

df[[
    "Customer_Satisfaction",
    "Customer_Feedback"
]].head()
```

```python
#Create Discount Category

df["Discount_Category"] = np.where(
    df["Max_Discount_Offered"] < 10,
    "Low Discount",
    np.where(
        df["Max_Discount_Offered"] < 20,
        "Medium Discount",
        "High Discount"
    )
)

df[["Max_Discount_Offered", "Discount_Category"]].sample(5)
```

```python
#Create Delivery Speed Category

df["Delivery_Speed"] = np.where(
    df["Delivery_Time"] <= 5,
    "Fast Delivery",
    "Standard Delivery"
)

df[[
    "Delivery_Time",
    "Delivery_Speed"
]].head()
```

```python
#Display Transformed Columns

print(df[[
    "Revenue_Category",
    "Customer_Feedback",
    "Discount_Category",
    "Delivery_Speed"
]].head())
```
